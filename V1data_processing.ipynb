## Process trigggers

Processing the triggers: we are using data from **zomicron** file and save them together in **triggers_total.csv**

### Join the triggers if are .root files

```python
folder = "/home/esanches/Projetos/tSNE/Virgo/zomicron"
exit_csv = "/home/esanches/Projetos/tSNE/Virgo/triggers_total.csv"
```

```python
files = [f for f in os.listdir(folder) if f.endswith(".root")]

first = True

for arq in files:
    print(f"Reading: {arq}")
    caminho = os.path.join(folder, arq)
    
    with uproot.open(caminho) as f:
        tree = f["triggers;1"]
        df = tree.arrays(library="pd")
        
    df.to_csv(exit_csv, mode="w" if first else "a", header=first, index=False)

    first = False
    del df
```

### break O3a and O3b from triggers

* To see the lenght, in CMD use: wc -l arquivename.csv
* To see the columns, in CMD use: head -n 1 arquivename.csv

```python
# from astropy.time import Time
import os
import pandas as pd
```

```python
run = 'O3b'
```

```python
gps_start_O3a = 1238166018
gps_end_O3a = 1253977218

gps_start_O3b = 1256655618
gps_end_O3b = 1269360018
```

```python
start = {'O3a' : gps_start_O3a, 'O3b' : gps_start_O3b}
end = {'O3a' : gps_end_O3a, 'O3b' : gps_end_O3b}

gps_start = start[run]
gps_end = end[run]
```

```python
arquive = "/home/esanches/Projetos/tSNE/Virgo/triggers_total.csv"
exit_arquive = "/home/esanches/Projetos/tSNE/Virgo/triggers_total_" + run + ".csv"
```

```python
total_triggers_window = 0
count = 0
first_write = True

for chunk in pd.read_csv(arquive, chunksize=10_000_000):
    print(f'Processing chunk: {count}')

    chunk['time'] = pd.to_numeric(chunk['time'], errors='coerce')
    
    mask = (chunk['time'] >= gps_start) & (chunk['time'] <= gps_end)
    filtered = chunk[mask]
    
    if not filtered.empty:
        total_triggers_window += len(filtered)
        
        filtered.to_csv(exit_arquive, mode='a', index=False, header=first_write)
        first_write = False
    
    count += 1
```
